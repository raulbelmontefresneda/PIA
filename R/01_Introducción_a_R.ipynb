## Instando librerías
La mayoría de las bibliotecas vienen preinstaladas en el servidor WSL. Sin embargo, si necesita instalar una nueva librería, necesita usar la función `install.packages()`.  Por ejemplo, puede instalar la biblioteca MASS (Modern Applied Statistics in S), dplyr, ggplo2, readr, haríamos lo siguiente:

```r
options(repos = c(CRAN = "https://cran.r-project.org"))
pkgs <- c("MASS","dplyr","ggplot2","readr")   # edita tu lista
need <- setdiff(pkgs, rownames(installed.packages()))
if (length(need)) install.packages(need, dependencies = TRUE)
lapply(pkgs, library, character.only = TRUE)
```

## Usando R como una calculadora
Por defecto, el resultado de la última línea se muestra en el cuaderno de Jupyter. Para ver otras líneas, use la funcion `print`.

```r
print(3+4) #Suma
print(7 %% 3) #Módulo 
print(6/2) #división
print(2^10) #potencia
print(sin(pi /2 )) # tenemos constantes como pi, y funciones trigonométricas
print(1.24e4)  # tambien se permite la notación científica de números.
print( 11  %%  3  ==  1 )
12>3
```

1. Si depositas 100 &euro; en un banco que paga un 5% de interés compuesto diario, ¿cuánto tendrás en el banco al cabo de un año? Recuerda que, en la clase de matemáticas de secundaria,
$$ A = P(1 + r/n)^{n m} $$

Donde:
* A = Capital final.
* P = Capital inicial.
* r = Tasa de interés anual.
* n = Número de capitalizaciones por periodo (ejemplo: 365 si es diario).
* m = Número de años

2. ¿Cuál es la distancia (euclidiana) entre dos puntos en (1, 5) y (4, 8)? Recordemos que la distancia euclidiana entre dos puntos (x1, y1) y (x2, y2) viene dada por el teorema de Pitágoras:
   $$ \sqrt{(x_1 - x_2)^2 + (y_1 - y_2)^2} $$
