## Tipos de datos
En cualquier lenguaje de programación, existen los conceptos de "tipos de datos" y "estructuras de datos". Esto se debe a que los programas manipulan datos, y cada tipo de datos requiere una manipulación diferente. Por ejemplo, los números se tratan de forma diferente a los caracteres (o cadenas de caracteres), y los números individuales se tratan de forma diferente a las listas de números (vectores) o matrices de números (arrays).

Los siguientes son algunos tipos de datos R simples:
* numeric
* character
* logical (TRUE o FALSE)
* complex    

Los tipos de datos básicos se pueden combinar para formar estructuras de datos:
* Vector
* Matrix
* String
* Data frame

Ejemplos:

```r
class(c(1,3,2,8.4)) #This is a vector

class(matrix(c(1,3,2,4,5,6),nrow=2,ncol=3)) #This is a matrix

"This is a string!"

data.frame(c("This is a string","This is another string"),matrix(c(1:6),nrow=2,ncol=3))
```

Algo importante a señalar es que los Data frames pueden combinar distintos tipos de datos.

## Creación de Objetos
### c - Concatenate
Acabamos de ver este comando en acción. El comando 'c' combina objetos por concatenación.
Este ejemplo crea un vector con 3 valores.

```r
c ( 5 , 6 , 7 )
```

Podemos añadir más valores al vector usando c (concatenar) resultando otro vector con más valores

```r
c(c(5,6,7),8,9)
```

Normalmente asignaremos los vectores con alguna variable con nombre. Por lo que generalmente usaremos un código más parecido a este:

```r
v1<-c(5,6,7)
v2<-c(v1,8,9)
print(v1)
print(v2)
```

### rbind
Si quisiéramos crear una matriz, podríamos usar el comando `matrix` como se indica arriba

```r
matrix(c(1,3,2,4,5,6),nrow=2,ncol=3)
```

O podemos crear vectores y combinarlos con `rbind` (rbind viene de "row bind" "enlace de filas")

```r
v1<-c(1,2,5)
v2<-c(3,4,6)
m1<-rbind(v1,v2)
m1
class(m1)
```

Observe que R asigna automáticamente nombres a las filas (v1 y v2).

### cbind
También tenemos cbind que viene column bind (enlace de columnas) y que nos puede servir para crear matrices o ampliarlas, como es el caso del siguiente ejemplo:

```r
m1<-matrix(c(1,2,3,4),nrow=2,ncol=2)
m1
```

```r
m2<-cbind(m1,c(5,6))
m2
```

## Como Acceder a la Documentación de R
Aunque la mejor forma de aprender es programando, por eso les pondré unos cuadernos de ejercicios para que puedan practicar/jugar con los comandos de R.
Podemos usar Google para encontrar la documentación sobre R, aunque el propio R dispone de algunas funciones que pueden seernos útiles para programar:
* __`?`:__ Si conoce el nombre del comando que del que desea conocer su uso, puede simplemente escribir `?nombre_del_comando` en una celda de código y ejecutarlo. Por ejemplo, para saber el uso del comando help pondríamos `?help`.
* __`help()`:__ Pasaremos como parámetro la función de la que queremos leer su documentción. Es similar a `?` 
* __`example()`:__ Muestra ejemplo de uso de la función que pasaremos por parámetro. Por ejemplo para ver ejemplos de uso de `sep` pondríamos: `example(seq)`
* __`apropos()`:__ Nos sirve para buscar funciones que contiene la cadena pasada por parámetro. Por ejemplo àpropos("seq") devuelve toas las funciones que contienen la cadena "seq"
* __`??`:__ Para una búsqueda más amplia utilizando coincidencias “difusas”, utilice ??“precedido” al término de búsqueda. Por ejemplo `??seq`


```r
?help
```

```r
help(seq)
```

```r
example(seq)
```

```r
apropos("seq")
```

```r
??seq
```
