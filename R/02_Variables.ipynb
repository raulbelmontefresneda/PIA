# Introducción a R
## Declaración y asignación de variables
En R **no se declara** una variable antes de usarla.  
La variable **se crea al asignarle un valor**.  
Los nombres de las variables han de cumplir ciertas reglas:
* Deben iniciar con letra o .
* Pueden contener letras, números y _
* Sin espacios ni caracteres especiales
* Evitar palabras reservadas

```r
x <- 3 # creación y asignación con <- 
y = 5  # Tambien es válido con = 

z <- x * y   # para la división  x/y
z
```

```r
z<-x/y
z
```

```r
z<-x**y # también valdría x^y  para las potencias.
z
```

### Creación y uso de  secuencias numéricas.

__Parámetros:__  
- **from**: valor inicial.
- **to**: valor final.
- **by**: incremento o decremento entre elementos.
- **length.out**: número total de elementos a generar.
- **along.with**: ajusta la longitud para que coincida con otro objeto.

__Sintaxis__  
`seq(from, to, by)`  
`seq(from, to, length.out = n)`

El operador : crea una secuencia de enteros consecutivos. Es una forma abreviada y muy eficiente para el caso específico de secuencias simples sin cambiar el paso.  
  `from:to`

__Comportamiento:__  
- Secuencias crecientes o decrecientes según el signo de **by**.
- Si el paso no llega al valor final exacto, la secuencia se corta antes de superarlo.
- Cuando se usa **length.out**, distribuye los valores de forma uniforme entre inicio y fin.
- **along.with** resulta útil para crear índices con la misma longitud que otro vector.

__Usos comunes:__  
- Construir rangos de valores.
- Controlar iteraciones en bucles.
- Crear secuencias para indexar datos o generar mallas de trabajo.


```r
1:10
```

```r
seq(2, 12, 3)
```

```r
seq(from=0, to=pi, by=pi/4)
```

```r
seq(from=0, to=pi, length.out = 5)
```

```r
x <- seq(1, 10)
#con [] podemos filtrar
x[x %% 3 == 0] # Ejemplo: filtrando por los elementos que sean múltiplos de 3
```

```r
x <- seq(1, 10)
x * 2  # así multiplicaríamos a todos los elementos de la sequencia = Escalar * Matriz
```

```r
x<-1:5
y<-1:5
x * y  # devuelve una secuencia donode se multiplica xi * yi 
```

```r
x <- seq(0, 2*pi, length.out = 5) # secuencia de 0 a 2pi con 5 elementos
print(x)
sin(x) # devuelve una secuncia de 5 elementos con cada elemento = sin(xi)
```
