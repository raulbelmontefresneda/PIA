## Ejercicio 1
* Adivina un número aleatorio generado por tu programa entre 1 y 10
* Por cada intento del usuario hay que indicar el número de intentos que lleva y si el número es mayor o menor
* Modificar el programa para que haya un número máximo de intentos

```python
import random

secreto = random.randint(1, 20)
adivinado, intentos = False, 0
MAX_INTENTOS = 3

numero = int(input('Dime tu número: '))
while numero!=secreto and intentos < MAX_INTENTOS:
    intentos += 1
    if numero < secreto:
        print('Es mayor')
    else:
        print('Es menor')
    numero = int(input('Dime tu número: '))

if numero == secreto:
    print('¡Lo adivinaste! \nNúmero de intentos:', (intentos+1))
else:
     print('¡Lo siento has perdido! \nNúmero secreto:', secreto)
```

# Ejercicio 2
* Realiza un programa que lea de teclado la nota de un examen de cada uno de los alumnos de una clase y calcule la media, la nota más alta, la más baja y el número de alumnos presentados al examen. 
* La entrada de datos se acaba cuando se lee una nota negativa.

```python
# Inicializaciones
suma = 0  # Inicialización del acumulador para la suma de los valores
cont = 0  # Contador que cuenta el número de elementos leídos
max = 0  # Inicialización del valor máximo por defecto
min = 10  # Inicialización del valor mínimo por defecto

# Lectura y validación de la entrada de datos
numero = int(input("Introduce una nota: "))
while numero > 10:
    print("Error. El número debe ser menor o igual que 10")
    numero = int(input("Introduce una nota: "))

while numero >= 0:
    suma = suma + numero
    cont = cont + 1
    if numero > max:
        max = numero
    if numero < min:
        min = numero
    # Lectura y validación de la entrada de datos
    numero = int(input("Introduce una nota: "))
    while numero > 10:
        print("Error. El número debe ser menor o igual que 10")
        numero = int(input("Introduce una nota: "))

if cont > 0:
    print("El valor máximo es", max)
    print("El valor mínimo es", min)
    print("La media es", suma / cont)
else:
    print("No se ha introducido ningún número positivo")
```

## Ejercicio 3
* Realizar un programa que juegue a piedra, papel y tijera, con las siguientes especificaciones:
    * Habrá un jugador que jugará contra la máquina.
    * El número de partidas a realizar será una constante en el programa. El juego se repetirá todas las veces que indique dicha constante.
    * Se solicitará al jugador que introduzca su jugada PIEDRA, PAPEL o TIJERA, mediante el siguiente mensaje:
¿PIEDRA, PAPEL O TIJERA?
    * El programa sacará aleatoriamente su jugada (piedra, papel o tijera).
* El programa indicará quién ha ganado según las siguientes reglas:
    * La piedra aplasta la tijera (Gana la piedra.)
    * La tijera corta el papel (Gana la tijera.)
    * El papel envuelve la piedra (Gana el papel.)
    * En caso de empate se señalará EMPATE

```python
import random

# Se indica con una constante el número de partidas que se va a jugar
N_PARTIDAS = 4

# Se imprime por consola
print("*****Se realizan", N_PARTIDAS, "partidas*****")

# Bucle para jugar las partidas indicadas en la constante
for i in range(N_PARTIDAS):
    # Pregunta por la jugada del usuario y la almacena en la variable
    # jugada_usuario (jugada del usuario)
    jugada_usuario = input("¿PIEDRA, PAPEL O TIJERA?: ").lower()

    # Genera aleatoriamente la jugada del ordenador y la almacena en la variable jProg
    jugada_programa = random.randrange(0, 3)

    # Utilizamos un condicional para convertirlo a texto
    # 0 -> piedra, 1-> papel, 2-> tijera
    if jugada_programa == 0:
        jugada_programa = "piedra"
    elif jugada_programa == 1:
        jugada_programa = "papel"
    else:
        jugada_programa = "tijera"

    # Alternativa
    # lista = ["piedra", "papel", "tijera"]
    # jugada_programa = lista[random.randrange(0, 3)]

    # Imprime la jugada del ordenador
    print("Programa saca", jugada_programa)

    if ((jugada_usuario == "piedra" and jugada_programa == "tijera") or
            (jugada_usuario == "tijera" and jugada_programa == "papel") or
            (jugada_usuario == "papel" and jugada_programa == "piedra")):
        print("*****GANA USUARIO*****")
    elif ((jugada_programa == "piedra" and jugada_usuario == "tijera") or
          (jugada_programa == "tijera" and jugada_usuario == "papel") or
          (jugada_programa == "papel" and jugada_usuario == "piedra")):
        print("*****GANA PROGRAMA*****")
    else:
        print("*****EMPATE*****")
```

## Ejercicio 4
* Realizar un programa que juegue al Ventiuno. Para este juego se tirarán dos dados. Los puntos se calcularán como la suma de las tiradas de los dos dados. Aunque al Ventiuno se puede jugar con varios jugadores, para simplificar jugaremos únicamente con dos.

Las reglas son las siguientes:
   * Cada jugador juega solo una vez por partida.
   * El jugador puede tirar los dados tantas veces como lo desee y puede plantarse en cualquier momento mientras la puntuación no sea mayor de 21. Cada tirada se va sumando al anterior para establecer la puntuación.
   * Si un jugador alcanza una puntuación mayor de 21 pierde inmediatamente, por lo que el otro jugador es automáticamente el ganador.
   * En el caso de que ninguno de los dos se haya pasado de 21, el participante cuyo total esté más cerca de los 21 puntos, pero no supere dicha cifra, gana. Si los dos jugadores alcanzan la misma cifra, entonces tendremos un empate.
   * El número de partidas a jugar será un valor definido como una constante en el programa. 

* A partir de las especificaciones establecidas en el enunciado y del ejemplo proporcionado: 
   * Programar la jugada correspondiente al jugador 1 hasta que se plante o la puntuación sea mayor que 21, en este último caso el jugador 1 habrá perdido.
   * Programar también la jugada correspondiente al jugador 2 e imprimir por consola el ganador (o el empate) a partir de la puntuación obtenida por ambos jugadores.
   * Modificar el ejercicio para que realice varias partidas según lo indicado en el enunciado.

```python
import random

# Constante que define el numero de partidas que se van a jugar
N = 3  

# Bucle para jugar N partidas
for i in range(1, N+1):
    # Inicializacion de variables del Jugador 1 para cada partida
    # Acumulador para calcular la puntuación del jugador 1
    puntos_jugador1 = 0  
    seguir_jugador1 = True  # Indica si se debe seguir tirando o no

    # Jugador 1
    while seguir_jugador1:
        print("PARTIDA", i, "- JUGADOR 1")
        # Calculamos los puntos obtenidos aleatoriamente
        dado1 = random.randint(1, 6)
        dado2 = random.randint(1, 6)

        puntos_jugador1 = puntos_jugador1 + dado1 + dado2;
        print("Ha obtenido en los dados", dado1, ",", dado2)
        print("Su puntuación acumulada es", puntos_jugador1)
        if puntos_jugador1 > 21:
            seguir_jugador1 = False
        else:
            seguir_jugador1 = input("¿Desea seguir tirando (Sí|No)? ").lower()\
                              != "no"
        print()

    # Inicializacion de variables del Jugador 2 para cada partida
    puntos_jugador2 = 0  # Acumulador para calcular la puntuación del jugador 2
    seguir_jugador2 = True  # Indica si se debe seguir tirando o no
    if puntos_jugador1 > 21:
        seguir_jugador2 = False

    while seguir_jugador2:
        print("PARTIDA", i, "- JUGADOR 2")
        # Calculamos los puntos obtenidos aleatoriamente
        dado1 = random.randint(1, 6)
        dado2 = random.randint(1, 6)
        puntos_jugador2 = puntos_jugador2 + dado1 + dado2;
        print("Ha obtenido en los dados", dado1, ",", dado2)
        print("Su puntuación acumulada es", puntos_jugador2)

        if puntos_jugador2 > 21:
            seguir_jugador2 = False;
        else:
            seguir_jugador2 = input("¿Desea seguir tirando (Sí|No)? "
                                    "").lower() != "no"
        print()

    # Resultados
    if puntos_jugador1 > 21:
        print("***********GANA EL JUGADOR 2***********")
    elif puntos_jugador2 > 21:
        print("***********GANA EL JUGADOR 1***********")
    elif puntos_jugador1 > puntos_jugador2:
        print("***********GANA EL JUGADOR 1***********")
    elif puntos_jugador2 > puntos_jugador1:
        print("***********GANA EL JUGADOR 2***********")
    else:
        print("***********EMPATE***********")
```
