# Introducción a las funciones
* Permiten al programador reorganizar el código dividiéndolo en partes más pequeñas
* Las funciones no se ejecutan en el momento de su definición sino cuando se llaman desde otra parte del código
* Razones para dividir un código en funciones: organización del código, reutilización, mantenimiento, abstracción...

# Definición de una función
* Antes de llamar a la función hay que definirla (declararla)
* Una función se compone de las siguientes partes: cabecera, documentación(opcional) y cuerpo
* Reglas de estilo: los nombres de las funciones se escriben en minúscula y las palabras deben estar separadas por guiones
* Importante: una función no hace nada hasta que no se invoca

```python
# Definición de una función
def es_par(n):
    if n%2 == 0:
        print("Es par")
    else:
        print("Es impar")

#Llamada a una funcion
es_par(8)

numero = 5
es_par(numero)
```

# Parámetros
* Para pasar información a la función utilizamos los parámetros
* Puede haber múltiples parámetros en una función
* Los parámetros son variables que se declaran en la cabecera
* No es necesario que las variables con los que invocamos a la función tengan el mismo nombre que los parámetros

```python
def procesar_numeros(num1, num2, num3):
    """
    Esta función toma tres números como entrada e imprime:
    - La suma de los tres números si el primer número es positivo
    - El producto de los tres numeros si es negativo
    """
    if num1 > 0:
        print(f"La suma de los tres números es: {num1 + num2 + num3}")
    else:
        print(f"a multiplicación de los tres números es: {num1 * num2 * num3}")

# Ejemplo de uso de la función
numero1 = float(input("Introduce el primer número: "))
numero2 = float(input("Introduce el segundo número: "))
numero3 = float(input("Introduce el tercer número: "))

procesar_numeros(numero1, numero2, numero3)
```

# Llamadas a funciones
* Cuando se llama a una función hay que dar valores a todos los parámetros
* Hay dos formas de hacerlo
    * Por posición, el orden debe ser el que está definido en la función
    * Por el nombre del parámetro, en este caso el orden no importa, aunque se suele utilizar el mismo 
* Dentro de una función se puede llamar a otras funciones e incluso puede llamarse a sí misma (recursividad)

```python
def procesar_numeros(num1, num2, num3):
    """
    Esta función toma tres números como entrada y realiza:
    - La suma de los tres números si el primer número es positivo
    - El producto de los tres numeros si es negativo
    
    Parámetros:
    - num1:float - Primer número
    - num2:float - Segundo número
    - num3:float - Tercer número

    Devuelve:
    - Nada
    
    """
    if num1 > 0:
        print(f"La suma de los tres números es: {num1 + num2 + num3}")
    else:
        print(f"a multiplicación de los tres números es: {num1 * num2 * num3}")

# Ejemplo de uso de la función
numero1 = float(input("Introduce el primer número: "))
numero2 = float(input("Introduce el segundo número: "))
numero3 = float(input("Introduce el tercer número: "))

procesar_numeros(numero1, numero2, numero3)

procesar_numeros(num3 = numero3, num2 = numero2, num1 = numero1)
```

# Valores por defecto en parámetros (parámetros opcionales)
* Se puede declarar un parámetro y darle un valor por defecto
* Los parámetros por defecto deben ir detrás de los que no lo son

```python
def saludar(nombre, saludo='Buenos días'):
    """
    Esta función saluda a la persona con un mensaje personalizado.
    """
    mensaje = f"{saludo}, {nombre}"
    return mensaje

# Ejemplos de uso de la función
saludo_defecto = saludar('Ana')
print(saludo_defecto)

saludo = saludar('Carlos', saludo='Hola')
print(saludo)
```

# Devolviendo valores
* Las funciones en Python pueden devolver uno o varios valores
* Si son varios se separan por comas y lo que se devuelve es una tupla (veremos en el siguiente tema lo que es)

```python
# Ejemplo de una función
def calcular_factorial(n):
    """ Esta función calcula el factorial de un número y lo imprime """
    if n == 0 or n == 1:
         factorial = 1
    else:
        factorial = 1
        for i in range(2, n + 1):
            factorial *= i
    return factorial

# Ejemplo de uso
numero = 5
resultado = calcular_factorial(numero)
print(resultado)
```

# Especificando el tipo de los parámetros
* En Python no hay que declarar el tipo de los parámetros
* Podemos especificar su tipo de la siguiente manera:
    * Con anotaciones en la cabecera de la función parámetro:tipo (es únicamente informativo)
    * Indicándolo en la documentación de la función :param nombre_parametro: tipo y descripción, ó @param nombre_parametro: tipo y descripción (es únicamente informativo)
    * Comprobando el tipo con las funciones type() e instance()

```python
def calcular_media(num1: float|int, num2: float|int) -> float:
    """
    Esta función calcula la media de dos números y devuelve su valor
    Parámetros: 
    :param num1: un número entero o real
    :param num2: un número entero o real
    :return media: un número real
    """
    if  ((type(num1)==int or type(num1)==float) and
        (type(num2)==int or type(num2)== float)):
        media = (num1 + num2) / 2
        return media
    else:
        raise TypeError("Los parámetros deben ser número enteros o flotantes")


# Ejemplo de uso de la función
numero1 = 10.5
numero2 = 7.2

media_resultante = calcular_media(numero1, numero2)
print(f"La media de {numero1} y {numero2} es: {media_resultante}")
```
