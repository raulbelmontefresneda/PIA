# Listas y tuplas anidadas
* En otros lenguajes se conocen como matrices
* Si utilizamos el operador = se copia la referencia en memoria de la lista
* Si utilizamos el operador .copy() se hace una copia superficial. Se copian las referencias de los elementos.
* Para hacer una copia profunda copiamos elementos uno a uno o  importando la librería copy llamamos al método deepcopy

```python
# Lista de listas
lista_de_listas = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

# Accediendo a elementos de la lista anidada
print(lista_de_listas[0])  # Imprime la primera lista
print(lista_de_listas[1][2])  # Imprime el tercer elemento de la segunda lista 
print(lista_de_listas[2][0])  # Imprime el primer elemento de la tercera lista

# Modificando elementos de la lista anidada
lista_de_listas[1][1] = 99
print(lista_de_listas) 

# Lo mismo se puede hacer con tuplas de tuplas, listas de tuplas y tuplas de listas
# Tupla de tuplas
tupla_de_tuplas = ((1, 2, 3), (4, 5, 6), (7, 8, 9))

# Lista de tuplas
lista_de_tuplas = [(1, 2, 3), (4, 5, 6), (7, 8, 9)]

# Tupla de listas
tupla_de_listas = ([1, 2, 3], [4, 5, 6], [7, 8, 9])

import copy

print("Lista original:", lista_de_listas)
# Utilizamos copy para crear una copia superficial de la lista anidada original
copia_superficial = copy.copy(lista_de_listas)

# Utilizamos deepcopy para crear una copia profunda de la lista anidada original
copia_profunda = copy.deepcopy(lista_de_listas)

# Modificamos un elemento en la lista original y en sus copias
lista_de_listas[0][0] = 99

# Imprimimos todas las listas para ver los cambios
print("Lista original después de modificar:", lista_de_listas)
print("Copia superficial:", copia_superficial)
print("Copia profunda:", copia_profunda)

#Recorrer listas de listas

# Recorremos la lista de listas utilizando un bucle for anidado
for lista in lista_de_listas:
    for elemento in lista:
        print(elemento, end=" ")
    print()  
```

# Diccionarios
* Son un conjunto mutable y heterogéneo de datos indexado por clave.
* Permite almacenar pares clave-valor de forma eficiente
* Se puede crear un diccionario utilizando llaves {} y especificando los pares clave-valor separados por comas. Ejemplo: mi_diccionario = {"nombre": "Juan", "edad": 30, "ciudad": "Madrid"}

```python
# Crear un diccionario vacío
mi_diccionario = {}

# Crear un diccionario con algunos elementos
mi_diccionario = {"nombre": "Juan", "edad": 30, "ciudad": "Madrid"}

# Acceder al valor asociado con la clave "nombre"
print('mi_diccionario["nombre"]', mi_diccionario["nombre"])  

# Modificar el valor asociado con la clave "edad"
mi_diccionario["edad"] = 35
print('mi_diccionario["edad"]', mi_diccionario["edad"])  

# Agregar una nueva entrada al diccionario
mi_diccionario["profesion"] = "Ingeniero"
print('mi_diccionario["profesion"] = "Ingeniero"', mi_diccionario)

# Eliminar la entrada asociada con la clave "ciudad"
del(mi_diccionario["ciudad"])
print('del(mi_diccionario["ciudad"]:', mi_diccionario)

# Verificar si una clave está presente en el diccionario
print("'nombre' in diccionario': ", 'nombre' in mi_diccionario) 
print("'peso' not in diccionario", 'peso' not in mi_diccionario)  
```

## Funciones

```python
# Creamos un diccionario de ejemplo
diccionario = {'a': 1, 'b': 2, 'c': 3}

# len(dic): Devuelve el número de claves en el diccionario
print("Número de claves en el diccionario:", len(diccionario))

# del(dic[clave]): Borra una clave específica del diccionario
del(diccionario['b'])
print("del(diccionario['b']): ", diccionario)

# Unión de diccionarios: No se puede usar el operador + para unir dos diccionarios
# Intentar usar el operador + para unir diccionarios resultará en un TypeError
# nuevo_diccionario = diccionario1 + diccionario2

# No se puede utilizar el slicing en diccionarios
# No es aplicable a diccionarios, ya que no tienen un orden definido.
```

## Métodos

```python
# Creación de un diccionario
diccionario = {'a': 1, 'b': 2, 'c': 3}

# Método `keys()`: Devuelve las claves del diccionario
print(diccionario.keys())  

# Método `values()`: Devuelve los valores del diccionario
print(diccionario.values())  

# Método `items()`: Devuelve los  pares clave-valor del diccionario
print(diccionario.items())  

# Iterar sobre las claves del diccionario
for clave in mi_diccionario:
    print(clave)

# Iterar sobre los valores del diccionario
for valor in mi_diccionario.values():
    print(valor)

# Iterar sobre los pares clave-valor del diccionario
for clave, valor in mi_diccionario.items():
    print(clave, valor)

# Método `get()`: Devuelve el valor asociado con la clave especificada
print(diccionario.get('a'))  
print(diccionario.get('d'))  

# Método `pop()`: Elimina y devuelve el valor asociado con la clave especificada
valor = diccionario.pop('b')
print(valor)
print(diccionario)  

# Método `update()`: Actualiza el diccionario con los pares clave-valor de otro diccionario
diccionario1 = {'a': 1, 'b': 2}
diccionario2 = {'c': 3, 'd': 4}
diccionario1.update(diccionario2)
print(diccionario1)  

# Método `copy()`: Crea una copia superficial del diccionario
copia_diccionario = diccionario.copy()
copia_diccionario['d'] = 4

print(id(diccionario))
print(id(copia_diccionario))

# Método `clear()`: Elimina todos los elementos del diccionario
diccionario.clear()
print("diccionario.clear():", diccionario)
```

## Operadores

```python
# Definimos dos diccionarios
diccionario1 = {'a': 1, 'b': 2, 'c': 3}
diccionario2 = {'b': 2, 'c': 3, 'a': 1}
diccionario3 = {'a': 1, 'b': 2, 'c': 4}
diccionario4 = {'d': 4, 'e': 5, 'f': 6}

# Comparamos los diccionarios usando los operadores relacionales
print(diccionario1 == diccionario2)  # True, ambos diccionarios tienen los mismos pares clave-valor
print(diccionario1 == diccionario3)  # False, diccionario3 tiene un valor diferente para la clave 'c'
print(diccionario1 != diccionario4)  # True, los diccionarios son diferentes

# Los operadores >, >=, <, y <= no están definidos para diccionarios en Python.
# Intentar usar estos operadores generará un TypeError.

# El operador = copia la referencia en memoria del diccionario igual que en listas

# Asignamos un diccionario a otro. Es la misma referencia al mismo diccionario.
diccionario2 = diccionario1
print("id(diccionario1):", id(diccionario1))
print("id(diccionario2):", id(diccionario2))

# Modificamos una clave en el diccionario1 y se refleja en el diccionario2
diccionario1['b'] = 4
print("diccionario1['b'] = 4:", diccionario1)
print("diccionario2:", diccionario2)

# Copiamos un diccionario con el método copy(). Son diccionarios diferentes. La copia es superficial
diccionario3 = diccionario1.copy()
print("id(diccionario1):", id(diccionario1))
print("id(diccionario3:", id(diccionario3))
```

# Conjuntos
* Python también incluye un tipo de dato para conjuntos.
* Un conjunto es una colección no ordenada y sin elementos repetidos.
* Los conjuntos también soportan operaciones matemáticas como la unión, intersección, diferencia, y diferencia simétrica.

```python
cesta = {'manzana', 'naranja', 'manzana', 'pera', 'naranja', 'plátano'}
print(cesta)               


a = set('abracadabra')
print(a) 

A = {1, 2, 3}
B = {3, 4, 5}

print(A|B, A&B, A-B, A^B)

print(
    "Unión:", A.union(B),
    "| Intersección:", A.intersection(B),
    "| Diferencia (A - B):", A.difference(B),
    "| Diferencia Simétrica:", A.symmetric_difference(B)
)
```

# Pilas
* Se puede implementar facilmente una pila con una lista
* Donde el último elemento añadido es el primer elemento el salir "Last in, first out".
* Para introducir un elemento en la pila, se utiliza append().
* Para sacar un elemento de la cima de la pila, se utiliza pop() 

```python
pila = [3, 4, 5]
pila.append(6)
pila.append(7)
print(pila)
pila.pop()
print(pila)
```

# Colas
* Agregar y sacar del final de la lista es rápido, pero insertar o sacar del comienzo de una lista es lento (porque todos los otros elementos tienen que ser desplazados en uno).
* Por ello es conveniente utilizar  collections.deque el cual fue diseñado para añadir y quitar de ambos extremos más rápidamente.

```python
from collections import deque

queue = deque(["Eric", "John", "Michael"])
queue.append("Terry")     
queue.append("Graham")   
queue.popleft()               
queue.popleft()                

print(queue)
```
