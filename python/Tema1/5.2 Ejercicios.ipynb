# Ejercicio 1
* Modifica el ejercicio de la semana pasada para trabajar con clases.
* Debes crear una clase Liga y una clase Equipo. Los equipos se almacenarán en la liga con un diccionario.
* El programa deberá registrar los resultados de los partidos y mantener actualizada la clasificación de los equipos participantes.
* El sistema almacenará la información relevante de cada equipo, incluyendo su nombre, partidos jugados, goles marcados, goles recibidos y puntos acumulados.
* Se sumará tres puntos si el equipo gana, 1 si hay un empate y 0 si pierde.
* Se comprobará que son valores válidos con el setter.
* Para generar la tabla se introducirá la información de los partidos jugados por teclado.
* Los datos  de entrada son los siguientes:
    * Nombre del equipo local
    * Goles del equipo local
    * Nombre del equipo visitante
    * Goles del equipo visitante
* Se recibirán en una línea con el siguiente formato: Local, GolesLocal, Visitante, GolesVisitante

# Ejercicio 2
* Crea una clase Diccionario que permita almacenar palabras. Habrá también una clase Palabra que incluirá la palabra, su definición y un enlace a la RAE donde se trate dicha palabra. El diccionario siempre debe estar ordenado alfabéticamente, por lo que habrá que tener cuidado a la hora de insertar las palabras en él.
* Añade un método que permita agregar palabras ordenadamente y otro para hacer búsquedas en el diccionario.
* Necesitarás un programa de prueba para comprobar que todo funciona correctamente.
