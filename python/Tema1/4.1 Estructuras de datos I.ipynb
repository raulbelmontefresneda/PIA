## Funciones y métodos para los tipos básicos
* Los tipos básicos en Python str, int, float, bool y complex incorporan métodos y funciones que permiten trabajar con ellos para realizar diferentes operaciones.
* Algunas de ellas son:
    * int: bin, oct, hex, bit_length, to_bytes.
    * float: is_integer().
    * int y float: abs, divmod, max, min, pow, round.
    * complex: real, imag, abs
    * str: len, lower, upper, capitalize, strip, find, split, startswith, endswith, isdigit, isalpha, join, index
* https://docs.python.org/es/3/library/stdtypes.html#string-methods
* https://docs.python.org/es/3/library/stdtypes.html#additional-methods-on-integer-types
* https://docs.python.org/es/3/library/stdtypes.html#additional-methods-on-float
* En la librería strings tenemos algunas constantes interesantes
   * https://docs.python.org/3/library/string.html

```python
# Para int
# bin, oct y hex pasan un entero a binario, octal y hexadecimal respectivamente
x = 10
resultado = bin(x)  # '0b1010'
print("Obtenemos el número:", x, "en binario, octal y hexadecimal:", bin(x), oct(x), hex(x), 
      "y el número de bits para representarlo", x.bit_length())

#Para float
# is_integer comprueba si un float es equivalente a un número entero
y = 5.0
print("¿El número", y, "es entero?", y.is_integer())  

y = 5.5
print("¿El número", y, "es entero?", y.is_integer()) 

# Para int y float
# abs calcula el valor absoluto de un número
x, y = -7, -3.14
print("Valor absoluto de", x, abs(x), " y de", y, abs(y))  

# div mov obtiene el cociente y el resto de una división
x = 10
y = 3
print("Cociente y resto de", x, y, divmod(x, y)) 

# max y min obtiene el máximo y el mínimo de una lista de números
print("max:", max(1, 5, 3.7))  
print("min:",min(1, 5, 3.7))  

# pow obtiene la potencia de un número
print("pow: ", pow(2, 3))  

# round redondea un número
x = 3.14159
print("round:", round(x, 2))  

# Para complex
# real e imag obtienen la parte real e imaginaria de un número complejo
z = 3 + 4j
print("real:", z.real, "imag:", z.imag, "abs:", abs(z))

# Para str
# len obtienen el número de caractéres de una cadena
texto = "Hola Mundo"
print("El número de caractéres del texto", texto, "es: ", len(texto))

# lower, upper y capitalize convierten una cadena a mayúsculas, minúsculas y pone la primera letra 
# de la cadena en mayúsculas respectivamente
print(texto.lower() ,texto.upper(), texto.capitalize())  

# strip quita los caracteres en blanco a la izquierda y la derecha de la cadena
texto = " Hola Mundo   "
print("Eliminamos los caractéres a la izquierda y la derecha:", texto.strip())

# find buscar una subcadena dentro de otra cadena. Devuelve el índice de la primera aparición 
# de la subcadena, o -1 si la subcadena no se encuentra
# index es parecido a find. Lanza un ValueError si no encuentra la subcadena.
print("Buscamos 'Mundo' en la cadena con find: ", texto.find("Mundo"))
print("Buscamos 'Mundo' en la cadena con index: ", texto.index("Mundo"))

# split divide una cadena en una lista de subcadenas, utilizando un delimitador especificado. 
texto = "Hola Mundo Python"
print("Dividimos", texto, "utilizando como delimitadores los blancos", texto.split())

# startswith y endswith devuelve True si la cadena empieza o acaba con la subcadena, repectivamente
texto = "Hola Mundo"
print("El texto empieza por 'Hola':", texto.startswith("Hola"), "o acaba:", texto.endswith("Hola"))

# isdigit devuelve True si el texto está formado por dígitos
texto = "12345"
print("EL texto está formado por dígitos:", texto.isdigit())

# isalpha devuelve True si el texto está formado por letras del alfabeto
texto = "Hola"
print("El texto está formado por letras:", texto.isalpha())  

# join Este método se usa para unir elementos de una lista u otro iterable en una sola cadena, 
# utilizando un delimitador especificado:
print("-".join(texto))
```

# Listas
* Son secuencias mutable y ordenada de elementos indexados por posición.
* Los elementos pueden ser de diferentes tipos y se pueden cambiar después de la creación.
* Se definen utilizando corchetes [].
    * lista = \[elemento1, elemento2, elemento3]

```python
# Crear una lista
mi_lista = ["uno", False, 3.4, "cuatro", 5.5, True]

# Acceder a un elemento de la lista
print(mi_lista[3])

# Los índices negativos cuentan desde el final de la lista
print(mi_lista[-1])

# Obtenr una sublista
print(mi_lista[1:3])

#Crear una lista con variables
a, b, c = 3, 4, 5
mi_lista2 = [a, b, c]

#Unir dos listas para crear una nueva
mi_lista3 = mi_lista + mi_lista2
print(mi_lista3)

# Con un bucle for 
mi_lista4 = []
for i in range(5):
    mi_lista4 += [i]
print(mi_lista4)

# Con el operador *
mi_lista5 = mi_lista4 * 2
print(mi_lista5)

# Comprobar si un elemento está en la lista
# Operadores in y not in
print(4 in mi_lista5)

# Ejemplo de recorrido bucle for each
# Suma los elementos de una lista
suma = 0
for elemento in mi_lista5:
    suma = suma + elemento
print(suma)
```

## Funciones

```python
# Definimos una lista de ejemplo
lista = [5, 2, 8, 1, 3]

# Imprimir la lista
print(lista)  

# Obtener la longitud de la lista
longitud = len(lista)
print("len(lista):", longitud) 

# Eliminar todos los elementos de la lista
del(lista[:])
print("del(lista[:]):", lista)  

# Definimos una nueva lista
lista = [5, 2, 8, 1, 3]

# Eliminar el elemento en la posición 2 de la lista
del(lista[2])
print("del(lista[:]):", lista) 

# Eliminar los elementos entre la posición 1 y 3 de la lista
del(lista[1:3])
print("del(lista[1:3]):", lista) 

# Definimos una nueva lista desordenada
lista = [5, 2, 8, 1, 3]

# Devolver una nueva lista ordenada con los elementos de la lista original
lista_ordenada = sorted(lista)
print("sorted(lista):", lista_ordenada)  
```

## Métodos

```python
lista = [5, 2, 8, 1, 3]
lista.append(6)  # Agrega el número 6 al final de la lista. Es más eficiente que lista + [valor]
print("lista.append(6):", lista)

lista.insert(2, 7)  # Inserta el número 7 en la posición 2
print("lista.insert(2, 7): ", lista)

lista.remove(3)  # Elimina el número 3 de la lista
print("lista.remove(3): ", lista)

elemento_eliminado = lista.pop(2)  # Elimina y devuelve el elemento en la posición 2
print(elemento_eliminado)
print("lista.pop(2): ", lista)

# Si no se proporciona ninguna posición devuelve el último
elemento_eliminado2 = lista.pop()  # Elimina y devuelve el elemento en la posición 2
print(elemento_eliminado2)
print("lista.pop(): ", lista)

otra_lista = [4, 5, 6]
# Usamos el método extend para agregar los elementos de otra_lista al final de la lista inicial
lista.extend(otra_lista)
print("lista.extend(otra_lista): ", lista)  

# Obtenemos el índice de la primera aparición del valor 20
indice = lista.index(4)
print(indice) 
print("lista.index(4): ", indice)

# Devuelve el número de veces que aparece el valor 1
print("lista.count(1):", lista.count(1))

# Borramos todos los elementos de la lista
lista.clear()
print("lista.clear():", lista)
```

## Operadores
* Operadores de comparación (==, !=, >, >=, <=, <)
    * Se compara elemento a elemento hasta que se puede dar una respuesta   
* Operadores de asignación (=) y copia de listas
    * El operador = se utiliza para asignar una lista a una variable.
    * Al asignar una lista a otra variable, se copia la referencia a la memoria.
    * Para copiar el contenido de una lista a otra sin compartir la misma referencia, se puede usar el método copy().
    * La función id() devuelve la posición de memoria de una variable, -> se usa para verificar si dos variables apuntan al mismo objeto en la memoria.

```python
lista1 = [1, 2, 3]
lista2 = [1, 2, 4]
print("lista1 < lista2", lista1 < lista2) 
print("lista1 > lista2", lista1 > lista2) 

lista1 = [1, 2, 3]
lista2 = [3, 2, 1]
print("lista1!= lista2", lista1 != lista2) 

# Asignamos una lista a otra. Es la misma lista
lista3 = lista1
print(id(lista1))
print(id(lista3))

#Modificamos lista1[1] y se modifica en lista3
lista1[1] = 3
print(lista1)
print(lista3)

# Copiamos una lista con el operador copy ->  son diferentes listas
lista4 = lista1.copy()
print(id(lista1))
print(id(lista4))
```

# Tuplas
* Son secuencias inmutables y ordenadas de elementos indexados por posición
* IMPORTANTE: No se pueden modificar los elementos, si se hace la tupla cambia.
* Se definen utilizando paréntesis. Los paréntesis son opcionales
    * tupla = (elemento1, elemento2, elemento3)
* Para convertir una lista en tupla, se utiliza tuple(lista)
* Para convertir una tupla en lista, se utiliza list(tuple)

```python
# Crear una tupla
mi_tupla = ("hola", True, 3.14, "mundo", 42, False)
print(id(mi_tupla))

# Si añadimos un elemento la tupla cambia ya que es inmutable. 
mi_tupla = mi_tupla + ("hola",)
print(id(mi_tupla))

# Acceder a un elemento de la tupla
print(mi_tupla[3]) 

# Los índices negativos cuentan desde el final de la tupla
print(mi_tupla[-1])  

# Obtener una subtupla
print(mi_tupla[1:3])

# Crear una tupla con variables
x, y, z = 10, 20, 30
mi_tupla2 = (x, y, z)

# Unir dos tuplas para crear una nueva
mi_tupla3 = mi_tupla + mi_tupla2
print(mi_tupla3)  

# Con el operador *
mi_tupla4 = mi_tupla3 * 2
print(mi_tupla4)  

# Comprobar si un elemento está en la tupla
# Operadores in y not in
print(4 in mi_tupla4)  

# Ejemplo de recorrido bucle for each
# Suma los elementos de una tupla
mi_tupla5 = (5, 6, 7, 8, 9, 10)
suma = 0
for elemento in mi_tupla5:
    suma += elemento
print(suma) 

# Convierte una tupla en una lista
mi_lista = list(mi_tupla5) 
print(mi_lista, type(mi_lista))

# Convierte una lista en una tupla
mi_tupla6 = tuple(mi_lista) 
print(mi_tupla6, type(mi_tupla6))
```

## Funciones

```python
# Definimos una tupla de ejemplo
tupla = (5, 2, 8, 1, 3)

# Imprimir la tupla
print(tupla)  

# Obtener la longitud de la tupla
longitud = len(tupla)
print("len(tupla):", longitud) 

# No se puede insertar o eliminar elementos de una tupla directamente,
# ya que las tuplas son inmutables

# Definimos una nueva tupla desordenada
tupla = (5, 2, 8, 1, 3)

# Devolver una nueva lista ordenada con los elementos de la tupla original
lista_ordenada = sorted(tupla)
print("sorted(tupla):", lista_ordenada)  
```

## Métodos

```python
# Definimos una tupla de ejemplo
tupla = (5, 2, 8, 1, 3)
print(tupla)

# No podemos utilizar el método append, insert, remove, pop, extend en tuplas,
# ya que son inmutables

# Podemos usar el método index en tuplas para encontrar el índice de un elemento
indice = tupla.index(8)
print("tupla.index(8):", indice)

# Podemos usar el método count en tuplas para contar cuántas veces aparece un elemento
cuenta = tupla.count(1)
print("tupla.count(1):", cuenta)

# Como las tuplas son inmutables, no existe un método clear para borrar todos los elementos.
```

## Operadores

```python
tupla1 = (1, 2, 3)
tupla2 = (1, 2, 4)

# Comparamos si tupla1 es menor que tupla2
print("tupla1 < tupla2:", tupla1 < tupla2)

# Comparamos si tupla1 es mayor que tupla2
print("tupla1 > tupla2:", tupla1 > tupla2)

tupla1 = (1, 2, 3)
tupla2 = (3, 2, 1)

# Comparamos si tupla1 es diferente de tupla2
print("tupla1 != tupla2:", tupla1 != tupla2)

# Asignamos una tupla a otra. Se copia la referencia en memoria de la lista
tupla3 = tupla1
print(id(tupla1))
print(id(tupla3))

# Intentamos modificar un elemento de la tupla1. Esto dará un error ya que las tuplas son inmutables.
# tupla1[1] = 3
# print(tupla1)

# No existe copy para tuplas
```

# Algunas funciones y métodos interesantes sobre listas y tuplas
* La función enumerate() retorna una tupla que contiene un contador (desde start, cuyo valor por defecto es 0) y los valores obtenidos al iterar sobre iterable.
* El método choice() del módulo random en Python se utiliza para seleccionar aleatoriamente un elemento de una secuencia.
* El método choices() del módulo random en Python se utiliza para seleccionar múltiples elementos aleatorios de una secuencia.

```python
import random
# Ejemplo enumerate
estaciones = ['Primavera', 'Verano', 'Otoño', 'Invierno']
print(list(enumerate(estaciones)))

for indice, valor in enumerate(estaciones):
    print(valor, indice)

# Ejemplo choice
print("La estación elegida es", random.choice(estaciones))

# Ejemplo choices
print("La estación elegida es", random.choices(estaciones, k=2))
```
