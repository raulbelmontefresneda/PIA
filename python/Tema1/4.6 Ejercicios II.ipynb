# Ejercicio 1.
* Realiza un programa que lea las filas y columnas que tiene una matriz.
* A continuación leer por teclado los números que forman la matriz fila a fila.
* Según se vaya leyendo cada fila calcular el producto de los números de las filas de la matriz y añadir ese número a cada una de las filas
* Imprimir el resultado con el siguiente formato:
*   4 * 5 * 2 = 40
*   5 * 5 * 5 = 125
*   3 * 3 * 6 = 54 

```python
# Leer las dimensiones de la matriz
filas = int(input("Introduce el número de filas: "))
columnas = int(input("Introduce el número de columnas: "))

# Leer los elementos de la matriz fila por fila y calcular el producto de cada fila
matriz = []
for i in range(filas):
    fila = []
    producto = 1
    for j in range(columnas):
        numero = int(input("Introduce un número: "))
        fila.append(numero)
        producto *= numero                  
    fila.append(producto)
    matriz.append(fila)
                
# Imprimir los elementos de la fila con el producto calculado
for i in range(len(matriz)):
    for j in range(len(matriz[i])-2):
        print(matriz[i][j], ' * ', end='')
    print(matriz[i][len(matriz[i])-2], ' = ', matriz[i][len(matriz[i])-1])
```

# Ejercicio 2.
* Vamos a desarrollar un programa para jugar al bingo.
* El bingo es un juego que consiste en extraer bolas numeradas de un bombo que contiene 90 bolas). 
* Cada jugador tiene un cartón con números aleatorios distribuidos en filas y columnas.
* Un locutor va sacando bolas del bombo, anunciando los números en voz alta. Si un jugador tiene el número anunciado en su cartón, lo tacha. El juego continúa hasta que un jugador completa todos los números de su cartón.
* Cuando un jugador completa una línea horizontal en su cartón, debe anunciar "línea".
* Si un jugador completa todos los números de su cartón, debe anunciar "bingo" y se convierte en el ganador de la partida.
* Para simular el juego del bingo con un solo cartón, se utilizará un cartón que consta de una cuadrícula con tres filas y cinco números en cada fila.

```python
import random


def generar_carton():
    carton = []
    for i in range(3):
        fila = []
        for j in range(5):
            num = random.randint(1, 90)
            while num in carton:  # Evitar números repetidos en el cartón
                num = random.randint(1, 90)
            fila.append(num)
        carton.append(fila)
    return carton


# Función para verificar si hay línea en una fila
def hay_linea(carton):
    for fila in carton:
        linea = True
        for elemento in fila:
            if elemento != 0:
                linea = False
        if linea:
            return True
    return False


def hay_bingo(carton):
    for fila in carton:
        for elemento in fila:
            if elemento != 0:
                return False
    return True


# Función para jugar al bingo con un cartón
def jugar_bingo(carton):
    numeros_extraidos = []  # Lista para almacenar los números extraídos
    bingo = False
    linea = False
    while not bingo:
        numero_extraido = random.randint(1,
                                         90)  # Extrae un número aleatorio del 1 al 90
        numeros_extraidos.append(numero_extraido)
        print(f"Número extraído: {numero_extraido}")

        # Tacha el número en el cartón si está presente

        for i in range(len(carton)):
            for j in range(len(carton[i])):
                if carton[i][j] == numero_extraido:
                    carton[i][j] = 0

        imprimir_carton(carton)  # Imprime el cartón actualizado

        if not linea:
            if hay_linea(carton):
                print("¡LÍNEA!")
                linea = True

        if hay_bingo(carton):
            print("¡BINGO!")
            bingo = True


# Función para imprimir un cartón de bingo
def imprimir_carton(carton):
    print("Cartón de Bingo:")
    for fila in carton:
        print(fila)


# Función principal
carton = generar_carton()
imprimir_carton(carton)
jugar_bingo(carton)
```

# Ejercicio 3
* Escribe una función en Python que tome como entrada un número entero entre 1 y 7 y devuelva el día de la semana correspondiente como un string. Utiliza un diccionario para mapear los números del 1 al 7 a los nombres de los días de la semana. Si el número está fuera del rango válido (1-7), la función debe devolver "Día inválido".

```python
def obtener_dia_semana(numero):
    dias_semana = {
        1: "Lunes",
        2: "Martes",
        3: "Miércoles",
        4: "Jueves",
        5: "Viernes",
        6: "Sábado",
        7: "Domingo"
    }
    if numero > 7 or numero < 1:
        return "Día inválido"
    else:
        return dias_semana[numero]
        
# Ejemplo de uso
numero = 3
print("El día correspondiente al número", numero, "es:", obtener_dia_semana(numero))
```

# Ejercicio 4
* Realiza un diccionario para guardar información sobre los empleados de una empresa. Cada empleado tendrá un código que será clave y su valor será otro diccionario con los datos del empleado: nombre, salario y departamento al que pertenece. Crear el diccionario con los datos de tres empleados. Imprime el diccionario.
* Añadir un nuevo empleado al diccionario.
* Borrar otro empleado
* Calcular el empleado con el salario máximo.

```python
# Diccionario inicial de empleados
empleados = {
    101: {"nombre": "Juan", "salario": 30000, "departamento": "Ventas"},
    102: {"nombre": "María", "salario": 35000, "departamento": "Marketing"},
    103: {"nombre": "Carlos", "salario": 32000, "departamento": "Ventas"}
}

# Imprimir información de todos los empleados
print("Información de empleados:")
for emp_id, info in empleados.items():
    print(f"Código: {emp_id}, Nombre: {info['nombre']}, Salario: {info['salario']}, Departamento: {info['departamento']}")

# Añadir un nuevo empleado
empleados[104] = {"nombre": "Ana", "salario": 40000, "departamento": "Desarrollo"}

# Borrar un empleado
del(empleados[103])

# Imprimir información actualizada de todos los empleados
print("\nInformación de empleados después de añadir y borrar:")
for emp_id, info in empleados.items():
    print(f"Código: {emp_id}, Nombre: {info['nombre']}, Salario: ${info['salario']}, Departamento: {info['departamento']}")

# Calcular empleado con salario máximo
max_salario = 0
codigo_max = -1
for emp_id, info in empleados.items():
    if (info["salario"] > max_salario):
        max_salario = info["salario"]
        codigo_max = emp_id

print("\nEmpleado con el salario máximo:")
print(f"Código: {codigo_max}, Nombre: {empleados[codigo_max]['nombre']}, Salario: ${empleados[codigo_max]['salario']}, Departamento: {empleados[codigo_max]['departamento']}")
```
