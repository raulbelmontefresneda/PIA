# Guía de estilo de Python
 * PEP 8 (Python Enhancement Proposal 8) - https://peps.python.org/pep-0008/
 * Describe las recomendaciones y convenciones de estilo para escribir código en Python.
 * Escrita por Guido van Rossum, y otros colaboradores
 * Destinada a promover la consistencia en el código Python
 * Traducida al español en el siguiente enlace: https://recursospython.com/guias-y-manuales/pep8es/

## Algunas recomendaciones
 * Indentación, cuatro espacios por nivel de indentación, evitar el uso de tabulaciones
 * Longitud de las líneas, 79 caracteres para el código y 72 para comentarios.
 * Para cortar líneas largas, se recomiendan paréntesis, corchetes o llaves, antes que la barra invertida ("\\")
 * Convenciones de nombres, snake_case para nombres de variables y funciones, CamelCase para nombres de clases
   
# Comentarios
* Un comentario comienza con un carácter de almohadilla (#) y termina al final de la línea.
* Son ignorados por el interprete de Python
* La función de los comentarios es describir el código de tal forma que resulte fácil de entender para cualquiera que lo lea un tiempo después (incluyéndote a ti mismo)
* Los comentarios tienen una importancia fundamental para garantizar la mantenibilidad del software desarrollado.
* Los docstring se utilizan para documentar la definición de clases, funciones o módulos. Se puede componer de varias líneas y se coloca entre tres comillas simples o dobles. Para acceder a ellos se utiliza el atributo __doc__

```python
# Ejemplo de comentarios
''' Este es un ejemplo
 para ilustrar un
 docstring'''
print(__doc__)
```

# Variables 
* Una variable es una posición de memoria a la que le asignamos un nombre y donde almacenamos datos
* No es necesario declarar el tipo de la variable
* Tipadon inferido y dinámico, el tipo de una variable se infiere automáticamente a partir del dato que se va a almacenar. Se determina en tiempo de ejecución y se puede cambiar a lo largo del programa
* Regla de estilo: minúsculas, subrayados para separar palabras (snake_case) 
* Algunas funciones interesantes, type(variable) e isinstance(variable, tipo)

```python
# Algunos ejemplos
var1 = 345
print("var1:", var1, type(var1))

#El tipo de una variable se puede cambiar, pero no se recomienda
var1 = "hola"
print("var1 con diferentes tipo:", var1, type(var1))
print("var1 es de tipo str (string): ", isinstance(var1, str))

var2 = None
print("var2:", var2)
```

# Constantes
* No hay constantes.
* Para indicar que una variable no debe cambiarse usamos mayúsculas y guiones bajos separando palabras.
* Por ejemplo, VENTA_MAX o TOTAL

```python
# En Python no hay constantes.
# Si queremos indicar que una variable no debe cambiarse, debemos usar mayúsculas
VENTA_MAX = 4_567
print(VENTA_MAX)

VENTA_MAX = "4"
print(VENTA_MAX)
```

# Identificadores
* Los identificadores pueden estar formados por letras (mayúsculas o minúsculas), dígitos y el carácter de subrayado _
* El primer carácter no puede ser un dígito
* Python distingue entre mayúsculas y minúsculas
* Los identificadores se utilizan para dar nombre a las variables, funciones, clases, etc
* Los nombres deben ser significativos y descriptivos, es decir tienen que revelar claramente lo que representan

# Tipos de datos

## Enteros (int)
 * El tamaño máximo depende solo de la memoria disponible
 * Ejemplos: 2020, 45678, -23, 345_267_923
 
## Reales (float)
 * Se almacenan en 64 bits
 * Presentan un problema de precisión, no se almacenan más de 15-16 digitos
 * Otros módulos para manejar float son decimal (precisión modificable) o fraction (aritmética exacta al basarse en fracciones) 
 * Hay valores especiales como nan (no es un número) o inf(infinito positivo y negativo)
 * Ejemplo: 4.567, 1e-4
   
## Complejos (complex)
 * Ejemplo: 6+8j

```python
import math

# Algunos ejemplos
var1 = 345_678_935
print("var1:", var1, type(var1))

var2 = "buenos días"
type(var2)
print("var2:", var2, type(var2))

var3 = 4.5
print("var3:", var3, type(var3))

var4 = 2 + 5j
print("var4:", var4, type(var4))

var5 = math.inf
print("var5:", var5, type(var5))

var6 = math.nan
print("var6:", var6, type(var6))

#Otras notaciones
print(4_567)
print(1e+6)
print(4_567.4)
print(1e-4)

# Problema de precisión en número flotantes
var7 = 9876543210987643.0
print(var7)
```

## Booleanos (bool)
* Valores: True / False (empiezan con mayúscula)
* True se corresponde con el valor int 1 y False con el valor int 0

# Casting (conversión de datos a otros tipos)
* Casting implícito: Ocurre automáticamente en ciertas ocasiones al operar diferentes tipos de datos. Por ejemplo, suma de float con int
* Casting explícito: Requiere el uso de funciones com str(), int(), float(), bool()

```python
# Casting explicito
var1 = 4
var2 = str(4)
print(var1, var2, type(var1), type(var2))
```

# Operadores

## Operadores de asignación
* Se utilizan para asignar valores a variables
* El operador básico es '=', por ejemplo var1 = 4
* Otros operadores de asignación (incluyen operación): +=, -=, *=, /=, %=, **=, //=

```python
# Podemos inicializar varias variables a la vez
var1 = var2 = var3 = "hola"
print("var1 var2 var3", var1, var2, var3)

# Múltiples asignaciones de forma paralela 
var4, var5, var6 = 4, "prueba", 2 + 5j
print("var4 var5 var6: ", var4, var5, var6)

# Otro ejemplo
var1, var2 = 10, 32
var1, var2 = 4, var1
print("¿Cuál es el resultado?")

# Para intercambiar el valor de dos variables
var1, var2 = var2, var1
var1 = 5
print(var1)

var1 += 4
print(var1)

var1 *= 6
print(var1)
```

## Operadores aritméticos
 * \+  suma o signo +
 * \-  resta o signo -
 * \*  multiplicación
 * /  división
 *  %  módulo (resto de la división)
 * // división entera -> se truncan los decimales del resultado
 *  \**  potencia

```python
# Ejemplos de operaciones
var1, var2 = 4, 8

var1 = var1 * var2
print(var1)

var1 = var1/3
print(var1)

var3 = var2//3
print(var3)

var3 = var2 % 3
print(var3)

# la división entera sigue el algoritmo de Euclides:  D=c*d + r , con 0 <= r  < d   
# Por lo tanto la división entera es c=0: porque 1 = 0*2 + 1
print(1//2)

#-1/2=-0.5 pero la división entera es c= ¿?
print(-1//2)  
```

## Operadores relacionales
* Se usan para comparar, devuelven True o False (un booleano)
    * == igual
    *  != distinto
    *  \> mayor
    *  \< menor
    *   \>= mayor o igual
    *    \<= menor o igual 
* Es posible establecer varias comparaciones a la vez, como en  var1 <= var2 <= var3
* Se pueden comparar cadenas de texto, para ello se compara el código Unicode carácter a carácter en las cadenas hasta que se determina cuál es mayor.

```python
var1 = 6
var2 = 5
var3 = 4
print (var1 > var2 > var3)

var4 = 'adios'
var5 = 'agua'
print (var4 < var5)
```

## Operadores lógicos o booleanos
* and: devuelve True si los dos operandos son True
* or: devuelve True si alguno de los operandos es True
* not: devuelve True si el operando es False y viceversa
* La evaluación es en "cortocircuito", se van evaluando los operandos uno a uno y se detiene la evaluación en cuanto se conoce el resultado.

```python
val1 = True
val2 = False
print (val1 and val2)
# Evalúa primero 4 > 5 como es False no evalúa más
print(4 > 5 and ((45678 % 5 // 4) >( 5 * 7656 /4)))
```

## Operadores de pertenencia
* Son in y not in. 
* Se utilizan para comprobar si un valor está presente en una secuencia (como una cadena, lista, tupla, etc.)

```python
s1 = "La lluvia en Sevilla es una pura maravilla"
print("lluvia" in s1)
```

## Operadores para cadenas de caracteres
* Concatenación (+), une dos cadenas
* Repetición (*), repite una cadena varias veces
* Longitud (len()), la función len() devuelve el número de caracteres de una cadena
* Índice y Slicing ([], [:]), permiten seleccionar caracteres individuales o subcadenas 
* Las cadenas en Python son inmutables, por lo que cualquier operación de slicing no modifica la cadena original, sino que crea una nueva.

```python
var1 = "En un lugar de la mancha "
var2 = "de cuyo nombre no quiero acordarme "
var3 = var1 + var2
print(var3) 
print ("Repitiendo var2 tres veces:", var2*3)
print("Longitud de var1:", len(var1))
print ("El cuarto elemento de var1 es: ", var1[3])
print("El penúltimo carácter de var1 es: ", var1[-2])
print("¿Qué hace esta operación? ", var1[3:8])
print("¿Qué hace esta operación?", var1[3:], var1[:4])
# Esta operación da error ya que las cadenas son inmutables
var1[4] = 5
```

## Operadores de bit
* Para obtener el binario de un número, podemos utilizar el operador (bin)
* Operador AND a nivel de bits (&)
* Operador OR a nivel de bits (|)
* Operador XOR a nivel de bits (^)
* Operador NOT a nivel de bits (~)
* Desplazamiento a la izquierda (<<) y desplazamiento a la derecha (>>)

```python
var1 = 876
print(var1, bin(var1))

var2 = var1 << 2
print(var2, bin(var2))

var3 = var1 >> 2
print(var3, bin(var3))

var4 = 567
print(var4, bin(var4))
var5 = 345
print(var5, bin(var5))

var6 = var4 & var5
print(var6, bin(var6))
```

## F-strings o cadenas formatedas
* Característica introducida en Python 3.6 que facilita la creación de cadenas con formato
* Para crear una f-string, hay que colocar una 'f' o 'F' antes de la cadena y utilizar llaves {} para encerrar las expresiones se desean evaluar

```python
nombre= "Juan"
apellidos= "Jiménez"

print(f'El nombre del alumno es {nombre} {apellidos}.')
```

## format()

```python
nombre= "Juan"
apellidos= "Jiménez"

print('El nombre del alumno es {} {}.'.format('Juan', 'Jiménez'))
```

## Especificadores de formato
* Patrones que se utilizan para controlar la presentación de los valores
* Ejemplos de formato de ancho y precisión:
    * {:.3f}: float con  3 decimales
    * {:<20}: Alineación a la izquierda en un campo de ancho 20
    * {:^5}: Alineación centrada en un campo de ancho 5
    * {:>8}: Alineación a la derecha en un campo de ancho 8{:.3s}: Mostrar solo los primeros 3 caracteres de una cadena
* Ejemplos de formato de enteros:
    * {:d}: int
    * {:02d}: int con dos ceros a la izquierda
* Ejemplos de formato de cadenas:
    * {:<15}: Alineación a la izquierda en un campo de ancho 15 para cadenas

```python
print(f'{4.6788:.3f}')
print(f'{"Nombre":<20} {"Apellidos":<20} {"Edad":>5}')
print(f'{"Juan Manuel":<20} {"Carrasco":<20} {24:>5}')
```

# Números pseudoaleatorios
* Hay que importar la librería random
* random.randrange(a, b): devuelve un entero aleatorio N tal que a <= N < b
* random.randint(a, b), devuelve un entero aleatorio N tal que a <= N <= b.
* random.random(): devuelve un float aleatorio N en tal que 0 <= N < 1
* random.uniform(a, b): devuelve un float aleatorio N tal que a <= N <= b para a <= b y b <= N <= a para b < a.

```python
import random
var1 = random.random()
var2 = random.randrange(1,5)
var3 = random.randint(1,5)
var4 = random.uniform(1,5)
print(var1, var2, var3, var4)
```
