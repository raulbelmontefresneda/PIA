# Ejercicio 
* Realizar un programa que juegue al tres en raya.
* Habrá dos tipos de fichas cruces "X" y ceros "0"
* El tablero estará compuesto por 9 casillas organizado en tres filas y tres columnas.
* Inicialmente el tablero estará vacío.
* Se jugará con dos jugadores, cada jugador tendrá un tipo de fichas CEROS y CRUCES.
* El programa permitirá que los jugadores coloquen una ficha alternativamente. S
* A continuación se evaluará la tirada.
* Si se llena el tablero y ninguno de los jugadores tiene un tres en raya, tendremos un empate.
* Si alguno consigue el tres en raya, gana.
* Utiliza los siguientes métodos:
    * inicializar_tablero()
    * imprimir_tablero(tablero)
    * colocar_ficha(tablero, fila, columna, ficha)
    * hay_tres_en_raya(tablero, ficha)

# Ejercicio 
* Realizar un programa para una liga de fútbol que permita registrar los resultados de los partidos y mantener actualizada la clasificación de los equipos participantes.
* Este sistema utilizará un diccionario en Python para almacenar la información relevante de cada equipo, incluyendo su nombre, partidos jugados, goles marcados, goles recibidos y puntos acumulados.
* Para generar la tabla se introducirá la información de los partidos jugados por teclado.
* Los datos son los siguientes:
    * Nombre del equipo local
    * Goles del equipo local
    * Nombre del equipo visitante
    * Goles del equipo visitante
* Se presentarán en una línea con el siguiente formato: Local, GolesLocal, Visitante, GolesVisitante
