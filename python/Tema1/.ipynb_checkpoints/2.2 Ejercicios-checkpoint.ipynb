## Ejercicio 3
* Escribe un programa que lea la edad y la altura de un usuario. 
* Debe comprobar que la altura no puede ser mayor que 250 cm y la edad no puede ser superior a 120 años. 
* Ambos números deben ser mayores que 0. 
* Si la edad o la altura es incorrecta emitirá un mensaje de error y volverá a pedir la edad al usuario.

# Ejercicio 4
* Hacer un programa que indique si un año leído por teclado es bisiesto.
* Un año es bisiesto si es múltiplo de 4, exceptuando los múltiplos de 100, que sólo son bisiestos cuando son múltiplos de 40.

# Ejercicio 5
* Desarrolla un programa que simule una competición de atletismo.
* El programa deberá mostrar los avances de cada uno de los competidores.
* Cada avance se calculará como un aleatorio entre 1 y 10.
* Se solicitará al usuario el largo de la pista que será un número entre 200 y 800.
* Se comprobará que la entradas para que se ajusten a los límites establecidos.
* Tendremos dos competidores, se pedirán sus nombres.
* En cada etapa de la carrera debe mostrar en la consola la posición actual de cada uno de los atletas, de la siguiente forma:
    * El atleta Noah Lyles se encuentra en la posición: 130
    * El atleta Kishane Thompson se encuentra en la posición: 156
* Cuando la carrera termina se indicará cuál es el atleta que ha ganado la carrera o si ha habido empate.
