# Ejercicios. Funciones

* Ponente: Ana María Blanco Benito

## Ejercicio 1
* Escribe una función en la que dados dos números enteros introducidos como parámetros, escriba los números primos que hay entre ambos.

## Ejercicio 2
* Escribe una función en la que dada una cadena de caractéres y una letra, devuelva el número de veces que aparece la letra en la cadena 

# Ejercicio 3
* Escribe una función en la que dada una cadena de caractéres, devuelva la cadena escrita del revés

## Ejercicio 4
* Escribe una función que dado un año como parámetro devuelva True si el año es bisiesto y False en caso contrario.Un año es bisiesto si es múltiplo de 4, exceptuando los múltiplos de 100, que sólo son bisiestos cuando son múltiplos de 400. Por ejemplo, el año 200 es bisiesto pero el 1900 no lo es.

## Ejercicio 5
* Realiza una función que calcule la letra que más se repite en una cadena introducida como parámetro.

## Ejercicio 6
* Realiza una función que calcule el dígito de control del NIF/NIE.
* Se introducirá como parámetro:
    * El NIF/NIE sin letra
    * Un booleano que será True, si lo que se ha introducido es un NIF y False si se ha introducido el NIE.
* Se podrá llamar a la función solo con el NIF/NIE sin letra. En este caso se considerará que se ha introducido un NIF
* Para ver cómo se calcula el dígito de control 
https://www.interior.gob.es/opencms/ca/servicios-al-ciudadano/tramites-y-gestiones/dni/calculo-del-digito-de-control-del-nif-nie/
