# Clases
* Una clase es la plantilla que utilizaremos posteriormente para crear una serie de objetos con características similares.
* Las clases agrupan en una misma estructura: los datos (atributos) y las operaciones a realizar con dichos datos (métodos).
* Por ejemplo, podemos definir una clase coche y los objetos (o instancias de clase) son los coches concretos, Hiunday i20 matrícula LJK4567, o Ford Focus matricula MJL5678.
* Para declarar una clase utilizamos la palabra class.
* A continuación se definen los atributos y luego los métodos
* Como regla de estilo nombres de las clases siguen el formato CamelCase

# Objetos
* Son instancias de clase. Ejemplares concretos.
* Para crear un objeto se escribe nombre_objeto = Clase(parámetros)
* Los parámetros que hay que introducir son los del constructor de la clase,

# Constructor
* Las clases tiene un método especial denominado constructor, que se utiliza declarar e inicializar los valores de los atributos cuando se crea un objeto.
* Este método tiene el nombre __init__
* Dentro del constructor se declaran los atributos de la clase poniendo self.nombre_atributo

# Métodos
* Un método es una función dentro de una clase.
* Es igual que cualquier función excepto que tiene que tener un parámetro. Este parámetro es self y es el primero del método.
* Para invocar un método escribimos los siguiente:
  nombre_objeto.nombre_método(parametros_sin_contar_self)
  

```python
#Declaración de la clase
class Alumno:
    #Constructor
    def __init__(self, nombre, edad, curso):
        # Atributos (datos)
        self.nombre = nombre
        self.edad = edad
        self.curso = curso
        self.notas = []

    # Métodos (operaciones - funciones)
    def agregar_nota(self, nota):
        self.notas.append(nota)

    def media(self):
        if len(self.notas) == 0:
            return 0
        else:
            return sum(self.notas) / len(self.notas)

# Ejemplo de uso
# Objeto alumno1
alumno1 = Alumno("Juan", 20, "DAW1")
alumno1.agregar_nota(75)
alumno1.agregar_nota(80)
alumno1.agregar_nota(85)

print("Nombre:", alumno1.nombre)
print("Edad:", alumno1.edad)
print("Curso:", alumno1.curso)
print("Notas:", alumno1.notas)
print("Media de las notas:", alumno1.media())
```

# Algunos métodos importantes
* __str__(self)
    * Por defecto si imprimo un objeto obtengo nombre_fichero.nombre_clase dirección_de_memoria, con este método se puede variar esta representación
* __eq__(self, otro_objeto)
    * Por defecto, comparar dos objetos con == , me da distinto aunque sus atributos sean iguales, ya que compara su posición en memoria. Puede cambiarse con este método.
* __repr__(sel)
    *  Define la representación del objeto.
* __lt__(self, other): Para definir el comportamiento del operador menor que (<).
* __le__(self, other): Para definir el comportamiento del operador menor o igual (<=).
* __gt__(self, other): Para definir el comportamiento del operador mayor que (>).
* __ge__(self, other): Para definir el comportamiento del operador mayor o igual (>=)

```python
class Alumno:
    def __init__(self, nombre, edad, curso):
        self.nombre = nombre
        self.edad = edad
        self.curso = curso
        self.notas = []

    def agregar_nota(self, nota):
        self.notas.append(nota)

    def media(self):
        if len(self.notas) == 0:
            return 0
        else:
            return sum(self.notas) / len(self.notas)

    def __str__(self):
        return f"Alumno: {self.nombre}, Edad: {self.edad}, Curso: {self.curso}, Notas: {self.notas}"

    def __eq__(self, other):
        return (self.nombre == other.nombre and
               self.edad == other.edad and
               self.curso == other.curso and
               self.notas == other.notas)

    def __lt__(self, other):
        if self.nombre == other.nombre:
            return self.media() < other.media()
        return self.nombre < other.nombre
    
    def __le__(self, other):
        if self.nombre == other.nombre:
            return self.media() <= other.media()
        return self.nombre <= other.nombre
    
    def __gt__(self, other):
        if self.nombre == other.nombre:
            return self.media() > other.media()
        return self.nombre > other.nombre
    
    def __ge__(self, other):
        if self.nombre == other.nombre:
            return self.media() >= other.media()
        return self.nombre >= other.nombre


# Ejemplo de uso
alumno1 = Alumno("Juan", 20, "DAW1")
alumno1.agregar_nota(7.5)
alumno1.agregar_nota(8.0)
alumno1.agregar_nota(8.5)

alumno2 = Alumno("Pedro", 21, "DAM2")
alumno2.agregar_nota(7)
alumno2.agregar_nota(7.5)
alumno2.agregar_nota(8)

print(alumno1)  # Imprime usando __str__

# Probando __eq__
alumno3 = Alumno("Juan", 20, "DAW1")
alumno3.agregar_nota(7.5)
alumno3.agregar_nota(8)
alumno3.agregar_nota(8.5)

print(alumno1 == alumno3)  
print(alumno1 == alumno2) 

print(alumno1 < alumno2)  
print(alumno1 < alumno3)  
print(alumno1 > alumno3)  

```

# Propiedades
* Es el equivalene a los getters y setters de Java.
* El objetivo es evitar que se puedan dar valores incorrectos a los atributos una vez el objeto se ha creado.
* El siguiente ejemplo ilustra su uso.

```python
class Alumno:
    def __init__(self, nombre, edad, curso):
        self.nombre = nombre
        self.edad = edad
        self.curso = curso
        self.notas = []

    @property
    def edad(self):
        return self.__edad

    @edad.setter
    def edad(self, value):
        if not 0 <= value <= 120:
            raise ValueError("La edad debe estar entre 0 y 120.")
        self.__edad = value

    def agregar_nota(self, nota):
        if not 0 <= nota <= 100:
            raise ValueError("La nota debe estar entre 0 y 100.")
        self.notas.append(nota)

    def promedio(self):
        if not self.notas:
            return 0
        return sum(self.notas) / len(self.notas)

    def __str__(self):
        return f"Alumno: {self.nombre}, Edad: {self.edad}, Curso: {self.curso}, Notas: {self.notas}"


# Ejemplo de uso
alumno1 = Alumno("Juan", 20, "DAW1")
alumno1.edad = 21  # Edad válida
#alumno1.edad = -1  # Esto dará un ValueError, edad fuera del rango
alumno1.agregar_nota(75)
alumno1.agregar_nota(80)
alumno1.agregar_nota(85)

print(alumno1)
```

# Atributos de solo lectura y atributos y métodos privados
* Para tener un atributo de solo lectura, que no se pueda modificar fuera de la clase, creo la property sin el setter
* Para tener un atributo privado, que no se pueda leer o modificar fuera de la clase, empiezo su nombre con dos guiones bajos.
* También se pueden tener métodos privados, que no se puedan ver fuera de la clase, empezando su nombre con dos guiones bajos

```python
class Alumno:
    def __init__(self, nombre, edad, curso):
        self.nombre = nombre
        self.edad = edad
        self.curso = curso
        self.notas = []

        self.__atributo_privado = "Hola, soy un atributo_privado"

    @property
    def edad(self):
        return self.__edad

    @edad.setter
    def edad(self, value):
        if not 0 <= value <= 120:
            raise ValueError("La edad debe estar entre 0 y 120.")
        self.__edad = value

    @property
    def atributo_solo_lectura(self):
        return "Hola soy un atributo solo lectura"

    def __metodo_privado(self):
        print("Soy un método privado")

    def agregar_nota(self, nota):
        if not 0 <= nota <= 100:
            raise ValueError("La nota debe estar entre 0 y 100.")
        self.notas.append(nota)

    def promedio(self):
        if not self.notas:
            return 0
        return sum(self.notas) / len(self.notas)

    def __str__(self):
        return f"Alumno: {self.nombre}, Edad: {self.edad}, Curso: {self.curso}, Notas: {self.notas}"


# Ejemplo de uso
alumno1 = Alumno("Juan", 20, "DAW1")
#alumno1.__metodo_privado()
#print(alumno1.__atributo_privado)
print(alumno1.atributo_solo_lectura)
#alumno1.atributo_solo_lectura = "Hola"

print(alumno1)
```

# Excepciones
* Cuando ocurre un evento inesperado podemos lanzar una excepción, por ejemplo cuando nos introducen un valor de un atributo incorrecto
* raise ValueError(mensaje) el mensaje es opcional, si el valor es incorrecto
* raise TypeError(mensaje), si el tipo es incorrecto
* Las excepciones se pueden manejar con try - except, que permite capturar errores o excepciones que podrían ocurrir durante la ejecución de un código y manejar esas excepciones sin que el programa se detenga abruptamente.
* También se pueden crear excepciones personalizadas, creando clases que hereden de Exception

```python
def suma_numeros(a, b):
    try:
        # Intentamos convertir ambos valores a enteros y sumarlos
        resultado = int(a) + int(b)
        print(f"La suma es: {resultado}")
    
    # Si alguno de los valores no es convertible a entero, se captura ValueError
    except ValueError:
        print("Error: Ambos valores deben ser números enteros.")
    
    # Si los argumentos son de tipo incorrecto, se captura TypeError
    except TypeError:
        print("Error: Ambos valores deben se enteros.")
    
    # Se maneja cualquier otra excepción no específica
    except Exception as e:
        print(f"Error inesperado: {e}")
    
    # Finalmente, siempre se ejecuta este bloque
    finally:
        print("Operación finalizada.")

# Pruebas
suma_numeros(10, "20")  # Funciona correctamente
suma_numeros(10, "abc")  # Provoca ValueError
suma_numeros([10], 20)  # Provoca TypeError
```

# Empaquetado de parámetros en funciones
* Se usa para pasar múltiples argumentos a una función. En la función se reciben como una sola tupla (empaquetado).
* Esto es útil cuando se trabaja con un número variable de argumentos o cuando se desea pasar múltiples valores a una función de manera más compacta.
* Se utiliza el operador \* para empaquetar tuplas.
* También se pueden empaquetar diccionarios con el operador \**

```python
def suma(*numeros):
    total = 0
    for num in numeros:
        total += num
    return total

# Empaquetar parámetros en una tupla y pasarlos a la función
resultado = suma(1, 2, 3, 4, 5)
print(resultado)  
```
