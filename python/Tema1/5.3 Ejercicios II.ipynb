# Ejercicio 1
* Crea una clase Estudiante con los siguientes atributos nombre, apellido y un diccionario con las notas de programación, lenguaje de marcas, bases de datos, entornos de desarrollo y sistemas.
* Crea las propiedades y los setters necesarios para comprobar que las notas se guarden en el rango correcto, si no es así genera una excepción
* Crea una clase Estadísticas que tendrá como atributo un estudiante y calculará la media, el máximo y el mínimo de sus notas.

```python
class Estudiante:
    def __init__(self, nombre, apellido):
        self.nombre = nombre
        self.apellido = apellido
        self.notas = {
            'programacion': 0,
            'lenguaje de marcas': 0,
            'bases de datos': 0,
            'entornos de desarrollo': 0,
            'sistemas': 0
        }

    @property
    def programacion(self):
        return self.notas['programacion']

    @programacion.setter
    def programacion(self, nota):
        self.validar_nota(nota)
        self.notas['programacion'] = nota

    @property
    def lenguaje_de_marcas(self):
        return self.notas['lenguaje de marcas']

    @lenguaje_de_marcas.setter
    def lenguaje_de_marcas(self, nota):
        self.validar_nota(nota)
        self.notas['lenguaje de marcas'] = nota

    @property
    def bases_de_datos(self):
        return self.notas['bases de datos']

    @bases_de_datos.setter
    def bases_de_datos(self, nota):
        self.validar_nota(nota)
        self.notas['bases de datos'] = nota

    @property
    def entornos_de_desarrollo(self):
        return self.notas['entornos de desarrollo']

    @entornos_de_desarrollo.setter
    def entornos_de_desarrollo(self, nota):
        self.validar_nota(nota)
        self.notas['entornos de desarrollo'] = nota

    @property
    def sistemas(self):
        return self.notas['sistemas']

    @sistemas.setter
    def sistemas(self, nota):
        self.validar_nota(nota)
        self.notas['sistemas'] = nota

    def validar_nota(self, nota):
        if nota < 0 or nota > 10:
            raise ValueError("La nota debe estar entre 0 y 10.")

    def __str__(self):
        return f"{self.nombre} {self.apellido} - Notas: {self.notas}"


class Estadisticas:
    def __init__(self, estudiante):
        self.estudiante = estudiante

    def calcular_media(self):
        return sum(self.estudiante.notas.values())/len(self.estudiante.notas)

    def calcular_maximo(self):
        return max(self.estudiante.notas.values())

    def calcular_minimo(self):
         return min(self.estudiante.notas.values())



def main():
    # Crear un estudiante
    estudiante = Estudiante("Carlos", "Pérez")
    estudiante.programacion = 9
    estudiante.lenguaje_de_marcas = 8
    estudiante.bases_de_datos = 7
    estudiante.entornos_de_desarrollo = 10
    estudiante.sistemas = 6

    print(estudiante)

    # Calcular estadísticas
    estadisticas = Estadisticas(estudiante)
    print(f"Media: {estadisticas.calcular_media():.2f}")
    print(f"Máximo: {estadisticas.calcular_maximo()}")
    print(f"Mínimo: {estadisticas.calcular_minimo()}")
    
main()
```

# Ejercicio 2.
* Crear una clase curso que permita almacenar todos los alumnos.
* Tendrá los siguientes atributos: curso, ciclo y una lista con los alumnos.
* Los métodos serán agregar estudiante y borrar estudiante y calcular medias, máximos y mínimos del curso

```python
class Estudiante:
    def __init__(self, nombre, apellido):
        self.nombre = nombre
        self.apellido = apellido
        self.notas = {
            'programacion': 0,
            'lenguaje de marcas': 0,
            'bases de datos': 0,
            'entornos de desarrollo': 0,
            'sistemas': 0
        }

    @property
    def programacion(self):
        return self.notas['programacion']

    @programacion.setter
    def programacion(self, nota):
        self.validar_nota(nota)
        self.notas['programacion'] = nota

    @property
    def lenguaje_de_marcas(self):
        return self.notas['lenguaje de marcas']

    @lenguaje_de_marcas.setter
    def lenguaje_de_marcas(self, nota):
        self.validar_nota(nota)
        self.notas['lenguaje de marcas'] = nota

    @property
    def bases_de_datos(self):
        return self.notas['bases de datos']

    @bases_de_datos.setter
    def bases_de_datos(self, nota):
        self.validar_nota(nota)
        self.notas['bases de datos'] = nota

    @property
    def entornos_de_desarrollo(self):
        return self.notas['entornos de desarrollo']

    @entornos_de_desarrollo.setter
    def entornos_de_desarrollo(self, nota):
        self.validar_nota(nota)
        self.notas['entornos de desarrollo'] = nota

    @property
    def sistemas(self):
        return self.notas['sistemas']

    @sistemas.setter
    def sistemas(self, nota):
        self.validar_nota(nota)
        self.notas['sistemas'] = nota

    def validar_nota(self, nota):
        if nota < 0 or nota > 10:
            raise ValueError("La nota debe estar entre 0 y 10.")

    def __str__(self):
        return f"{self.nombre} {self.apellido} - Notas: {self.notas}"

    def __eq__(self, otro):
        if not isinstance(otro, Estudiante):
            return False
        return self.nombre == otro.nombre and self.apellido == otro.apellido


class Curso:
    
    def __init__(self, nombre_curso, ciclo):
        self.nombre_curso = nombre_curso
        self.ciclo = ciclo
        self.estudiantes = []


    def agregar_estudiante(self, estudiante):
        self.estudiantes.append(estudiante)

    def borrar_estudiante(self, nombre, apellido):
        self.estudiantes.remove(Estudiante(nombre, apellido))    


    def calcular_maximo(self):
        if len(self.estudiantes) == 0:
            return -1
            
        maximo = 0
        for estudiante in self.estudiantes:
            maximo_estudiante =  max(estudiante.notas.values())
            if maximo_estudiante > maximo:
                maximo = maximo_estudiante
        return maximo

    def calcular_minimo(self):
        if len(self.estudiantes) == 0:
            return -1
            
        minimo = 10
        for estudiante in self.estudiantes:
            minimo_estudiante =  min(estudiante.notas.values())
            if minimo_estudiante < minimo:
                minimo = minimo_estudiante
        return minimo

    def calcular_media(self):
        if len(self.estudiantes) == 0:
            return -1
            
        total = 0
        for estudiante in self.estudiantes:
            total += sum(estudiante.notas.values())/len(estudiante.notas)
        return total


    def mostrar_estudiantes(self):
        if not self.estudiantes:
            print("No hay estudiantes en el curso.")
            return
            
        for estudiante in self.estudiantes:
            print(estudiante)

    def __str__(self):
        return f"Curso: {self.nombre_curso}, Ciclo: {self.ciclo}, Estudiantes: {len(self.estudiantes)}"



def main():
    # Crear un curso
    curso = Curso("Desarrollo de Software", "Primer Ciclo")

    # Crear estudiantes
    estudiante1 = Estudiante("Juan", "Pérez")
    estudiante1.programacion = 9
    estudiante1.lenguaje_de_marcas = 8
    estudiante1.bases_de_datos = 7
    estudiante1.entornos_de_desarrollo = 10
    estudiante1.sistemas = 6

    estudiante2 = Estudiante("María", "Gómez")
    estudiante2.programacion = 8
    estudiante2.lenguaje_de_marcas = 9
    estudiante2.bases_de_datos = 6
    estudiante2.entornos_de_desarrollo = 7
    estudiante2.sistemas = 10

    # Agregar estudiantes al curso
    curso.agregar_estudiante(estudiante1)
    curso.agregar_estudiante(estudiante2)

    # Mostrar estudiantes en el curso
    curso.mostrar_estudiantes()

    # Calcular estadísticas
    media = curso.calcular_media()
    maximo = curso.calcular_maximo()
    minimo = curso.calcular_minimo()
    
    # Mostrar estadísticas del curso
    print(f"Media del curso: {media:.2f}")
    print(f"Máximo del curso: {maximo}")
    print(f"Mínimo del curso: {minimo}")

    # Borrar un estudiante
    curso.borrar_estudiante("Juan", "Pérez")
    
    # Mostrar estudiantes después de la eliminación
    curso.mostrar_estudiantes()
    
main()
```

# Ejercicio 3.
* Para la gestión de una biblioteca vamos a crear una clase Libro con los siguientes atributos: titulo, autor, codigo ISBN, el año del publicación del libro y la cantidad de copias existentes en la biblioteca. Deberás métodos para prestar, devolver y comprobar la disponibilidad de ejemplares en la biblioteca.
* Escribir un programa que cree varios libros, solicite al usuario los valores de los campos por teclado e imprima por pantalla la información del libro.
* Gestiona alguno de los atributos con propiedades.
* Añade un atributo privado y otro solo lectura

```python
class Libro:
    def __init__(self, titulo, autor, codigo_ISBN, anio_publicacion, cantidad_ejemplares):
        self.titulo = titulo
        self.autor = autor
        self.codigo_ISBN = codigo_ISBN
        self.anio_publicacion = anio_publicacion
        self.cantidad_ejemplares = cantidad_ejemplares
        
        self.__atributo_privado = "Atributo privado"  # Atributo privado

    @property
    def cantidad_ejemplares(self):
        return self.__cantidad_ejemplares

    @cantidad_ejemplares.setter
    def cantidad_ejemplares(self, cantidad):
        if cantidad < 0:
            raise ValueError("La cantidad de ejemplares no puede ser negativa.")
        self.__cantidad_ejemplares = cantidad

    @property
    def atributo_lectura(self):
        return "Este es un atributo de solo lectura"

    def prestar(self):
        if self.cantidad_ejemplares > 0:
            self.cantidad_ejemplares -= 1
            return True
        return False

    def devolver(self):
        self.cantidad_ejemplares += 1
        return True

    def verificar_disponibilidad(self):
        if self.cantidad_ejemplares > 0:
           return True
        return False

    def __str__(self):
        return f"Título: {self.titulo}, Autor: {self.autor}, ISBN: {self.codigo_ISBN} \
                Año de publicación: {self.anio_publicacion}, Cantidad de ejemplares: {self.cantidad_ejemplares}"

# Solicitar al usuario los valores para crear un libro
titulo = input("Introduce el título del libro: ")
autor = input("Introduce el autor del libro: ")
codigo_ISBN = input("Introduce el código ISBN del libro: ")
anio_publicacion = int(input("Introduce el año de publicación del libro: "))
cantidad_ejemplares = int(input("Introduce la cantidad de ejemplares disponibles: "))

# Crear una instancia de la clase Libro
libro = Libro(titulo, autor, codigo_ISBN, anio_publicacion, cantidad_ejemplares)

# Imprimir la información del libro
print("\nInformación del libro:")
print(f"Título: {libro.titulo}")
print(f"Autor: {libro.autor}")
print(f"Código ISBN: {libro.codigo_ISBN}")
print(f"Año de publicación: {libro.anio_publicacion}")
print(f"Cantidad de ejemplares disponibles: {libro.cantidad_ejemplares}")

# Realizar algunas operaciones con el libro
libro.prestar()
print(libro)
print(libro.verificar_disponibilidad())
print(libro)
libro.devolver()
print(libro)
print(libro.verificar_disponibilidad())
print(libro)
```

# Ejercicio 4
* Crea una clase Biblioteca que permita gestionar los libros del ejercicio1, tendrás los siguientes métodos: agregar_libro, buscar_libro
  y contar el número de libros de la biblioteca que no tienen ningún ejemplar.
* Necesitarás un programa de prueba para comprobar que todo funciona correctamente.

```python
class Biblioteca:
    def __init__(self):
        self.libros = []

    def agregar_libro(self, titulo, autor, codigo_ISBN, anio_publicacion, cantidad_ejemplares):
        nuevo_libro = {
            "titulo": titulo,
            "autor": autor,
            "codigo_ISBN": codigo_ISBN,
            "anio_publicacion": anio_publicacion,
            "cantidad_ejemplares": cantidad_ejemplares
        }
        self.libros.append(nuevo_libro)
        print(f"Libro '{titulo}' agregado a la biblioteca.")

    def buscar_libro(self, titulo):
        for libro in self.libros:
            if libro["titulo"] == titulo:
                return libro
        return None

    def contar_libros_sin_ejemplares(self):
        contador = 0
        for libro in self.libros:
            if libro["cantidad_ejemplares"] == 0:
                contador += 1
        return contador

biblioteca = Biblioteca()
biblioteca.agregar_libro("El Quijote", "Miguel de Cervantes", "9788420412146", 1605, 5)
biblioteca.agregar_libro("Cien años de soledad", "Gabriel García Márquez", "9780307350476", 1967, 10)

print(biblioteca.buscar_libro("El Quijote"))
print(biblioteca.contar_libros_sin_ejemplares())
```
