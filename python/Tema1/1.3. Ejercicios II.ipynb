## Ejercicio 3
* Escribe un programa que pida un número en segundos y lo muestre en horas, minutos y segundos. La hora tendrá el formato HH:MM:SS

```python
segundos_totales = int(input("Introduce un número en segundos: "))

horas = segundos_totales // 3600
minutos = (segundos_totales % 3600) // 60
segundos = segundos_totales % 60

print(f"\n{segundos_totales} segundos equivalen a:")
print(f"Horas: {horas:02d}")
print(f"Minutos: {minutos:02d}")
print(f"Segundos: {segundos:02d}")
```

## Ejercicio 5
