# Paso de parámetros por valor y por referencia
* En los lenguajes de programación existen dos formas de pasarle una variable a una función que estamos llamando
    * Paso de parámetros por valor -> se hace una copia de la variable original. No se modifica la variable original
    * Paso de parámetros por referencia->  se opera sobre la variable original. Se modifica la variable original
* En Python el paso de parámetros es por valor
* Pero si el parámetro es de un tipo mutable se comporta como paso por referencia. En el siguiente tema se explicará la razón

```python
def modificar_valor(numero):
    print(f"Antes de la modificación: {numero}")

    # Intentar modificar el valor (esto no afectará al original fuera de la función)
    numero = 99

    print(f"Después de la modificación: {numero}")

# Crear un número fuera de la función
mi_numero = 42

# Llamar a la función con el número como argumento
modificar_valor(mi_numero)

# Imprimir el número después de llamar a la función
print(f"Después de la función: {mi_numero}")
```

# Recursividad
* Una función puede llamarse a sí misma
* Es muy útil para resolver problemas que pueden dividirse en subproblemas más pequeños y similares al problema original.
* Se establece un caso base que define la condición en la cual la función deja de llamarse a sí misma y simplemente devuelve un resultado directo

```python
def factorial_recursivo(n):
    """
    Calcula el factorial de un número de manera recursiva.

    Parámetros:
    - n: Número entero no negativo.

    Retorna:
        El factorial de n.
    """
    # Caso base: el factorial de 0 es 1
    if n == 0: 
        return 1
    # Caso recursivo: n! = n * (n-1)!
    else:
        return n * factorial_recursivo(n - 1)

# Ejemplo de uso
numero = 5
resultado_factorial = factorial_recursivo(numero)
print(f"El factorial de {numero} es: {resultado_factorial}")
```

# Ámbito de las variables
* Los parámetros y las variables locales internas a la función -> existen dentro de la función
* Las variables que se crean el programa principal -> existen desde que se crean hasta el final del programa
    * Pueden leerse en el interior de las funciones. No se recomienda
    * Con la palabra global pueden modificarse dentro de las funciones. No se recomienda

```python
variable = 10
def ejemplo_funcion(parametro_funcion):
    # Ámbito local: variable_local es válida solo dentro de esta función
    global variable
    variable_local = 5
    print("Dentro de la función:")
    print("parametro_funcion:", parametro_funcion)
    print("variable_local:", variable_local)
    print("variable (accesible dentro de la función):", variable)

    # Modificar variable global desde la función
    variable = 20

# Llamamos a la función con un parámetro
ejemplo_funcion(parametro_funcion=15)

# Intentamos acceder a variable_local fuera de la función (provocará un error)
# print("Fuera de la función - variable_local:", variable_local)

print("\nFuera de la función:")
print("variable_global (modificada dentro de la función):", variable)
```

# Expresiones lambda
* Se pueden crear pequeñas funciones anónimas con la palabra reservada lambda
* La sintaxis para crear una función lambda es lambda [parámetros]: expresión
* Están sintácticamente restringidas a una sola expresión. 

```python
# Definir una función lambda que suma dos números
suma = lambda x, y: x + y

# Usar la función lambda
resultado = suma(5, 7)
print(f"La suma es: {resultado}")
```

# Expresiones condicionales
* Una expresión condicional en Python es una forma compacta de escribir una condición que evalúa una expresión según sea verdadera o falsa, y devuelve un resultado en función de la evaluación. También se conoce como operador ternario
* La sintaxis es: valor_si_verdadero if condicion else valor_si_falso

```python
numero = 8
resultado = "Positivo" if numero > 0 else "Negativo"
print(resultado)
```

# Introducción a los módulos
* Un módulo es un fichero que contienen definiciones y declaraciones de variables, funciones, clases, etc
* El nombre de fichero es el nombre del módulo con la extensión .py
* Podemos crear nuestros propios módulos
* El objetivo es organizar el código para hacerlo más legible, reutilizable y fácil de mantener
* La biblioteca estándar de Python es un conjunto de módulos y paquetes que vienen incluidos con la instalación estándar de Python. Por ejemplo, random, math, datetime, decimal, fraction, sys

# Importar módulos
* Para utilizar un módulo en Python, primero hay que importarlo en tu programa
* Podemos importar módulos de dos formas:
    * _import \<nombre_modulo>_: importa el contenido completo del módulo, para acceder a los elementos del módulo _\<nombre_del_modulo>.\<elemento>_
    * _from \<modulo> import \<elemento>_: para importar elementos específicos, el elemento se llama directamente sin necesidad de prefijo. Si queremos incluir todos los elementos: _from \<modulo> import *_. 
* Se recomienda importar solo aquellos elementos que necesitemos.
* Las declaraciones import suelen estar al principio del programa
* Si queremos darle un alias al módulo importado por ejemplo porque el nombre es muy largo o porque puede haber conflictos de nombres utilizamos la palabra clave "as"

```python
import random as rd
from math import log10

aleatorio = rd.randint(1,5)

print(aleatorio)

print(log10(100))
```
